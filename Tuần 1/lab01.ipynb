## Bài 6. Phép toán và nhân ma trận

Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

## Bài 8. Autograd cơ bản

Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

Yêu cầu:


• Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.


• Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

## Bài 9. Gradient Descent tìm hệ số đa thức

Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

Yêu cầu:


• So sánh ba hệ số học được với [1, 2, 3].



• Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

## Bài 10. Giải hệ phương trình bằng Autograd

Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

Yêu cầu:

• Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.

• Nhận xét vì sao bài toán có thể có nhiều nghiệm.

Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.

```python
print("CÂU 2: Tensor A, B (3x4) - các phép toán và thống kê theo cột")


# Tạo A, B ngẫu nhiên kích thước 3x4 (dùng số dương để tránh chia cho 0)
A = torch.rand(3, 4) * 10 + 1   # giá trị trong khoảng [1, 11)
B = torch.rand(3, 4) * 10 + 1

print("A =\n", A)
print("\nB =\n", B)

A_plus_B = A + B
A_times_B = A * B          # nhân element-wise (không phải nhân ma trận)
A_div_B = A / B

print("\nA + B =\n", A_plus_B)
print("\nA * B (element-wise) =\n", A_times_B)
print("\nA / B =\n", A_div_B)

# Trung bình và độ lệch chuẩn theo CỘT -> tính theo chiều dim=0 (giảm theo hàng)
mean_col_A = A.mean(dim=0)
std_col_A = A.std(dim=0)
mean_col_B = B.mean(dim=0)
std_col_B = B.std(dim=0)

print("\nTrung bình theo cột của A:", mean_col_A)
print("Độ lệch chuẩn theo cột của A:", std_col_A)
print("\nTrung bình theo cột của B:", mean_col_B)
print("Độ lệch chuẩn theo cột của B:", std_col_B)
```

```python

print("CÂU 3: Autograd - y = 2x^2 + 5x + 3, tính dy/dx tại x = 2")


x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3

y.backward()  # tính đạo hàm dy/dx tự động

dy_dx_autograd = x.grad.item()
print(f"y tại x=2: {y.item()}")
print(f"dy/dx (autograd) tại x=2: {dy_dx_autograd}")

# Kiểm tra thủ công: dy/dx = 4x + 5 -> tại x=2: 4*2 + 5 = 13
dy_dx_manual = 4 * 2 + 5
print(f"dy/dx (tính tay) = 4x + 5 tại x=2: {dy_dx_manual}")
print(f"Khớp nhau? {'Có' if abs(dy_dx_autograd - dy_dx_manual) < 1e-6 else 'Không'}")
```

```python

print("CÂU 4: Thay đa thức bằng y = 2x^2 - 3x + 5, huấn luyện tìm hệ số")


# Sinh dữ liệu từ đa thức thật: y = 2x^2 - 3x + 5 (+ nhiễu nhỏ)
true_a, true_b, true_c = 2.0, -3.0, 5.0

x_data = torch.linspace(-5, 5, 200).reshape(-1, 1)
noise = torch.randn(x_data.shape) * 0.5
y_data = true_a * x_data**2 + true_b * x_data + true_c + noise

# Khởi tạo hệ số cần học: a, b, c (ngẫu nhiên, có requires_grad=True)
a = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
c = torch.randn(1, requires_grad=True)

optimizer = torch.optim.SGD([a, b, c], lr=0.001)
n_epochs = 3000

for epoch in range(n_epochs):
    optimizer.zero_grad()

    y_pred = a * x_data**2 + b * x_data + c
    loss = torch.mean((y_pred - y_data) ** 2)  # MSE

    loss.backward()
    optimizer.step()

    if (epoch + 1) % 500 == 0:
        print(f"Epoch {epoch+1}/{n_epochs} - Loss: {loss.item():.4f} "
              f"- a={a.item():.4f}, b={b.item():.4f}, c={c.item():.4f}")

print("\nKết quả cuối cùng ")
print(f"Hệ số thật:   a = {true_a}, b = {true_b}, c = {true_c}")
print(f"Hệ số học được: a = {a.item():.4f}, b = {b.item():.4f}, c = {c.item():.4f}")
```
