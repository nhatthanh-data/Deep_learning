```python
training_data = [[170, 65],
                 [155, 50],
                 [150, 45],
                 [175, 70],
                 [165, 55]]

print(training_data)
```

```python
import sympy as sym

a = sym.Symbol('a')
b = sym.Symbol('b')

Residual = 0

for i in range(len(training_data)):
    Residual += (training_data[i][0]*a + b - training_data[i][1]) **2


```

```python
Residual
```

```python
R_Diff_a = sym.diff(Residual, a)
R_Diff_a
```

```python
R_Diff_b = sym.diff(Residual, b)
R_Diff_b
```

```python
import numpy as np

A = [[170, 1],
     [155, 1],
     [150, 1],
     [175, 1],
     [165, 1]]

Y = [[65], [50], [45], [70], [55]]



At = np.transpose(A)

print(At)
```

```python

AtA = np.dot(At, A)
AtY = np.dot(At, Y)

print(AtA)
print(AtY)
inv_AtA = np.linalg.inv(AtA)
X = np.dot(inv_AtA, AtY)

print(X)

```

```python
# training data
arm = [40, 30, 20, 70, 60]
leg = [30, 80, 70, 40, 20]
gen = [1, -1, -1, 1, 1]
print(arm, leg, gen)

# differential

import sympy as sym

a = sym.Symbol('a')
b = sym.Symbol('b')
c = sym.Symbol('c')
R = 0
```

```python
# Residual
for i in range(0,5):
    R+=(arm[i]*a+leg[i]*b+c-gen[i])**2

R_a=sym.diff(R,a)
R_b=sym.diff(R,b)
R_c=sym.diff(R,c)
print("Result of differentiating with respect to a:", R_a, "= 0")
print("Result of differentiating with respect to b:", R_b, "= 0")
print("Result of differentiating with respect to c:", R_c, "= 0")
```

```python


import numpy as np

A = [[40, 30, 1],
     [30, 80, 1],
     [20, 70, 1],
     [70, 40, 1],
     [60, 20, 1]]

Z = [[1], [-1], [-1], [1], [1]]

At = np.transpose(A)
print(At)
```

```python
AtA = np.dot(At, A)
AtZ = np.dot(At, Z)

print(AtA)
print(AtZ)
inv_AtA = np.linalg.inv(AtA)
X = np.dot(inv_AtA, AtZ)
print(X)
```
